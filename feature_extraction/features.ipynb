```python
import numpy as np
import re
```

tokenization

```python
def sentence_split(text):
    return re.split(r'[.!?]+', text)
```

```python
def word_tokenize(text):
    return re.findall(r'\w+', text.lower())
```

```python
def avg_word_length(text):
    words = word_tokenize(text)
    if not words:
        return 0
    return np.mean([len(w) for w in words])
```

```python
def sentence_length_variance(text):
    sentences = sentence_split(text)
    lengths = [len(word_tokenize(s)) for s in sentences if s.strip()]
    if len(lengths) < 2:
        return 0
    return np.var(lengths)
```

```python
def avg_sentence_length(text):
    sentences = sentence_split(text)
    lengths = [len(word_tokenize(s)) for s in sentences if s.strip()]
    if not lengths:
        return 0
    return np.mean(lengths)
```

```python
def punctuation_ratio(text):
    punct = len(re.findall(r'[^\w\s]', text))
    total = len(text)
    return punct / total if total > 0 else 0
```

```python
def lexical_diversity(text):
    words = word_tokenize(text)
    if not words:
        return 0
    return len(set(words)) / len(words)
```

```python
# perplexity.py

import torch
from transformers import GPT2LMHeadModel, GPT2Tokenizer

tokenizer = GPT2Tokenizer.from_pretrained("gpt2")
gpt2_model = GPT2LMHeadModel.from_pretrained("gpt2")
gpt2_model.eval()

def compute_perplexity(text):
    encodings = tokenizer(text, return_tensors="pt")
    input_ids = encodings.input_ids

    with torch.no_grad():
        outputs = gpt2_model(input_ids, labels=input_ids)
        loss = outputs.loss

    return torch.exp(loss).item()
```

```python
# embedding_features.py

import numpy as np
from sentence_transformers import SentenceTransformer

em_model = SentenceTransformer('all-MiniLM-L6-v2')

def sentence_split(text):
    return [s.strip() for s in text.split('.') if s.strip()]

def embedding_variance(text):
    sentences = sentence_split(text)
    
    if len(sentences) < 2:
        return 0
    
    embeddings = em_model.encode(sentences)
    
    return np.var(embeddings)
```

```python
# stability.py

import numpy as np

def stability_score(predict_fn, text, perturb_fn):
    """
    predict_fn: your trained model prediction function
    perturb_fn: function that returns modified versions of text
    """

    variations = perturb_fn(text)

    preds = [predict_fn(v) for v in variations]

    return np.var(preds)
```
