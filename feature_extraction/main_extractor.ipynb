## extracting all the features

```python
import numpy as np
from tqdm import tqdm
```

```python
df_ai = df[df['label'] == 1].sample(400, random_state=42)
df_human = df[df['label'] == 0].sample(400, random_state=42)

df_small = pd.concat([df_ai, df_human]).sample(frac=1, random_state=42)
```

calling the function

```python

X = []
y = df_small['label'].values

for text in tqdm(df_small['full_text']):
    try:
        feats = extract_features(text)
        X.append(feats)
    except Exception as e:
        print("Error:", e)
        X.append([0]*7)  # fallback (since you have 7 features)

X = np.array(X)
```

```python
feature_df = pd.DataFrame(X, columns=[
    'perplexity',
    'sentence_length_variance',
    'embedding_variance',
    'avg_word_length',
    'punctuation_ratio',
    'lexical_diversity',
    'avg_sentence_length'
])

feature_df['label'] = y

feature_df.to_csv("../train/features_dataset.csv", index=False)

print("Features saved successfully!")
```
