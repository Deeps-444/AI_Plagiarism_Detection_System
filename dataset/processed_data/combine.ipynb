# cpmbining all the datasets in 1 csv with proper ratio

```python
import pandas as pd

df_human = pd.read_csv("human_clean.csv")
df_mixtral = pd.read_csv("mixtral_clean.csv")
df_argugpt = pd.read_csv("arguGPT_clean.csv")
```

```python
df_human.head("full_text")
```

standardizing

```python
df_ai = pd.concat([df_mixtral, df_argugpt], ignore_index=True)
df_ai = df_ai[["full_text", "label", "word_count"]]
```

```python
df_ai.head()
```

```python
df_human = df_human[df_human["word_count"] > 200]
df_ai = df_ai[df_ai["word_count"] > 200]
```

```python
n = min(len(df_human), len(df_ai))

df_human = df_human.sample(n, random_state=42)
df_ai = df_ai.sample(n, random_state=42)
```

```python
df_final = pd.concat([df_human, df_ai], ignore_index=True)
df_final = df_final.sample(frac=1, random_state=42).reset_index(drop=True)
```

```python
df_final.columns
```

```python
df_final = df_final.drop(columns=["Unnamed: 0"])
```

```python
df_final.columns
```

```python
df_final.to_csv("final_raw_ds.csv", index=False)
```
