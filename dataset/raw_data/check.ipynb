# persuade human essay dataset

```python
import pandas as pd
df = pd.read_csv("persuade2_.csv")
df["full_text"][0]
```

```python
df.columns
```

```python
print(len(df["full_text"][0].split()))
```

```python
df_clean = df[["full_text"]].copy()
df_clean["label"] = 0
df_clean["word_count"] = df_clean["full_text"].apply(lambda x: len(x.split()))
```

```python
df_clean.columns
```

```python
df_clean.to_csv("../processed_data/human_clean.csv")
```

 calculating number of essays with more than 200 words

```python
df["word_count_calc"] = df["full_text"].apply(lambda x: len(x.split()))
```

```python
below_200 = (df["word_count_calc"] <= 200).sum()
above_200 = (df["word_count_calc"] > 200).sum()
print("above 200: ", above_200)
print("below 200: ", below_200)
```
